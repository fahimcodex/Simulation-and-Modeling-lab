# Necessary Libraries

```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# Task 1: Scalar & Array Operations (Redo with scalar = 3.7 and 6 elements)

```python


scalar = 3.7
arr = np.array([1.2, 2.5, 3.8, 4.1, 5.9, 6.3])

# Mathematical functions
print(f"Scalar: {scalar}")
print(f"Array: {arr}")
print("sin:", np.sin(arr))
print("cos:", np.cos(arr))
print("tan:", np.tan(arr))
print("asin (clipped):", np.arcsin(np.clip(arr / np.max(arr), -1, 1)))
print("acos (clipped):", np.arccos(np.clip(arr / np.max(arr), -1, 1)))
print("atan:", np.arctan(arr))
print("exp:", np.exp(arr))
print("log:", np.log(arr))
print("abs:", np.abs(arr))
print("sqrt:", np.sqrt(arr))
print("remainder (div 2):", np.remainder(arr, 2))
print("round:", np.round(arr))
print("floor:", np.floor(arr))
print("ceil:", np.ceil(arr))
```

# Task 2 2X2 matrix

```python
# Create 2x2 matrix
matrix_2x2 = np.array([[1.5, 2.8], [3.2, 4.7]])

print("2x2 Matrix:\n", matrix_2x2)
print("\nsin:\n", np.sin(matrix_2x2))
print("\nexp:\n", np.exp(matrix_2x2))
print("\nlog:\n", np.log(matrix_2x2))
print("\nabs:\n", np.abs(matrix_2x2))
print("\nsqrt:\n", np.sqrt(matrix_2x2))
print("\nremainder (div 2):\n", np.remainder(matrix_2x2, 2))
print("\nround:\n", np.round(matrix_2x2))
print("\nfloor:\n", np.floor(matrix_2x2))
print("\nceil:\n", np.ceil(matrix_2x2))
```

# Task 4: Vector Functions on 2x3 Matrix

```python
# 2x3 matrix with float values
matrix_2x3 = np.random.uniform(1, 10, (2, 3))
print("2x3 Matrix:\n", matrix_2x3)

# Max/Min
print("Max:", np.max(matrix_2x3), "at index", np.argmax(matrix_2x3))
print("Min:", np.min(matrix_2x3), "at index", np.argmin(matrix_2x3))

# Length
print("Length (size):", matrix_2x3.size)

# Sorting descending
row_desc = -np.sort(-matrix_2x3, axis=1)
print("\nRow-wise descending sort:\n", row_desc)

entire_desc = -np.sort(-matrix_2x3.flatten())
print("Entire matrix descending sort:", entire_desc)

# Stats
print("\nSum:", np.sum(matrix_2x3))
print("Product:", np.prod(matrix_2x3))
print("Median:", np.median(matrix_2x3))
print("Mean:", np.mean(matrix_2x3))
print("Std Dev:", np.std(matrix_2x3))

# Bonus: Visualization
plt.figure(figsize=(10,8))
sns.heatmap(matrix_2x3, annot=True, cmap='viridis')
plt.title("Heatmap of 2x3 Matrix")
plt.show()
```
